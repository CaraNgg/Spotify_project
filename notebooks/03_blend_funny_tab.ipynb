```python
import sys
import pandas as pd
sys.path.insert(0, "/Users/carang/Documents/pycharm document projects/SpotifyProj/docs")
from spotify_loading import load_history, align_end_dates, filter_window

data = "/Users/carang/Documents/pycharm document projects/SpotifyProj/data/"

#fetch the data for both people
cara = load_history(data + "my_spotify_data2 copy.zip", "Cara")
ruby = load_history(data + "ruby_spotify_data copy.zip", "Ruby")

#concatenate both peoples data
both_clean = align_end_dates(pd.concat([cara, ruby], ignore_index=True))

# pick a window
window = input("Select a time frame (all, 30d, 6m, 12m, year): ").strip().lower()

if window == "year":
    year = int(input("Which year? "))
    both_clean = filter_window(both_clean, "year", year)
else:
    both_clean = filter_window(both_clean, window)

print("now in window:" + window)
```

```python
#longest song and artist streak
#DONT ADD
from blend_funny_tab import longest_phase

result = longest_phase(both_clean, "Cara", "Ruby")

for person in ["Cara", "Ruby"]:
    art = result[person]["artist"]
    song = result[person]["song"]
    print(f"🔥 {person}'s longest artist phase: {art['name']}, {art['days']} days in a row "
          f"({art['start'].strftime('%b %d, %Y')} – {art['end'].strftime('%b %d, %Y')})")
    print(f"   Longest song streak: {song['name']}, {song['days']} days in a row "
          f"({song['start'].strftime('%b %d, %Y')} – {song['end'].strftime('%b %d, %Y')})\n")

print(f"🏆 Longest phase winner: {result['winner']}")
```

```python
from blend_funny_tab import back_to_back_record

result = back_to_back_record(both_clean, "Cara", "Ruby")
#back to back to back plays
#DONT ADD

for person in ["Cara", "Ruby"]:
    r = result[person]
    hours = (r["end"] - r["start"]).total_seconds() / 3600
    print(f"🔂 {person} played {r['track']} by {r['artist']} {r['plays']} times in a row")
    print(f"   {r['start'].strftime('%b %d, %Y %I:%M %p')} → {r['end'].strftime('%I:%M %p')} "
          f"({hours:.1f} hours of just that song)\n")

print(f"🏆 Back-to-back winner: {result['winner']}")
```

```python
from blend_funny_tab import forgotten_favorite

result = forgotten_favorite(both_clean, "Cara", "Ruby")
#DONT ADD

for person in ["Cara", "Ruby"]:
    r = result[person]
    if r is None:
        print(f"👻 {person} has never abandoned a favorite. Loyal!\n")
        continue
    print(f"👻 {person}'s forgotten favorite: {r['track']} by {r['artist']}")
    print(f"   It was your #{r['peak_year_rank']} song of {r['peak_year']} ({r['peak_year_plays']} plays)")
    print(f"   You haven't played it since {r['last_played'].strftime('%b %Y')} "
          f"({r['days_since']} days ago)\n")
```

```python
#Top song in each season! (INDIV) DONT ADD
from blend_funny_tab import top_song_each_season

r = top_song_each_season(both_clean, "Cara", "Ruby")
emojis = {"Spring": "🌸", "Summer": "☀️", "Fall": "🍂", "Winter": "❄️"}

for person in ["Cara", "Ruby"]:
    print(f"{person}:")
    for season, song in r[person].items():
        if song:
            print(f"   {emojis[season]} {season}: {song['track']} by {song['artist']} ({song['plays']} plays)")
```

```python
#What three song does each person play the most after 12am? (INDIV)
#and the top three shared song after 12am HAHA (SHARED)
from blend_funny_tab import most_plays_after_twelve
from blend_funny_tab import top_shared_after_twelve

indiv = most_plays_after_twelve(both_clean, "Cara", "Ruby")
shared = top_shared_after_twelve(both_clean, "Cara", "Ruby")

for person in ["Cara", "Ruby"]:
    print(f"🌙 {person}'s late night playlist (12am–5am):")
    for i, song in enumerate(indiv[person], start=1):
        print(f"   {i}. {song['track']} by {song['artist']} ({song['plays']} plays)")
    print()

print("🌙 Your shared late night playlist:")
for i, song in enumerate(shared, start=1):
    print(f"   {i}. {song['track']} by {song['artist']} "
          f"(Cara: {song['Cara']} · Ruby: {song['Ruby']})")
```

```python
#What hour you guys are most likely to be listening at the same time,
#and what is each persons top song (SHARED)
from blend_funny_tab import top_shared_hour_top_song

r = top_shared_hour_top_song(both_clean, "Cara", "Ruby")
start = r["hour"]
end = (r["hour"] + 1) % 24
time_range = f"{start:02d}:00–{end:02d}:00"

print(f"🪞 You two are most likely to be listening at the same time between {time_range}.")
print(f"   That's when {r['Cara']['share']:.0%} of Cara's and {r['Ruby']['share']:.0%} of Ruby's listening happens.\n")

for person in ["Cara", "Ruby"]:
    x = r[person]
    print(f"   {person}'s most-played song at {r['hour']}:00, across all of their Spotify history:")
    print(f"   {x['track']} by {x['artist']} ({x['plays']} plays)\n")
```

```python
import sys
import pandas as pd
sys.path.insert(0, "/Users/carang/Documents/pycharm document projects/SpotifyProj/docs")
from spotify_loading import load_history, align_end_dates, filter_window

data = "/Users/carang/Documents/pycharm document projects/SpotifyProj/data/"

#fetch the data for both people
cara = load_history(data + "my_spotify_data2 copy.zip", "Cara")
ruby = load_history(data + "my_spotify_datahenkek.zip", "Ruby")

#concatenate both peoples data
both_clean = align_end_dates(pd.concat([cara, ruby], ignore_index=True))
```

```python
#MOST BALANCED ARTIST FOR ALL TIME
from blend_stats_tab import most_balanced_artist

result = most_balanced_artist(both_clean, "Cara", "Ruby")
print(f"Your most balanced artist: {result['artist']} "
      f"(Cara {result['Cara']:.1f} hrs, Ruby {result['Ruby']:.1f} hrs)")
```

```python
from blend_stats_tab import shared_taste_percent
#FOR ALL TIME
result = shared_taste_percent(both_clean, "Cara", "Ruby")
print(f"{result['Cara']:.0%} of Cara's listening goes to artists Ruby also listens to")
print(f"{result['Ruby']:.0%} of Ruby's listening goes to artists Cara also listens to")
```

```python
from blend_stats_tab import top_songs_overlap

result = top_songs_overlap(both_clean, "Cara", "Ruby")
pa = result["per_artist"]

print(f"{result['score']:.0%} of your top 5 songs match across your top shared artists\n")

for artist, info in pa.items():
    print(f"🎤 {artist}: {info['matches']} of {info['out_of']} match")

    # songs you both have in your top 5
    for song in info["shared_songs"]:
        print(f"   ✓ {song}  (both)")

    # songs only in one person's top 5
    for song in info["Cara_top"]:
        if song not in info["shared_songs"]:
            print(f"   • {song}  (Cara only)")
    for song in info["Ruby_top"]:
        if song not in info["shared_songs"]:
            print(f"   • {song}  (Ruby only)")

    print()
```

```python
#discovery style
from blend_stats_tab import discovery_style
result = discovery_style(both_clean, "Cara", "Ruby")

for person in ["Cara", "Ruby"]:
    info = result[person]
    print(f"How {person} discovered their top {info['artists_found']} artists:")
    for category, share in info["how"].items():
        print(f"   {category}: {share:.0%}")
    print(f"   Shuffle was on for {info['shuffle_on']:.0%} of discoveries\n")

    print("   🔍 Top artists you went looking for:")
    for artist, hrs in info["top_looked_for"]:
        print(f"      {artist} ({hrs:.1f} hrs)")

    print("   📻 Top artists that came to you:")
    for artist, hrs in info["top_came_to_you"]:
        print(f"      {artist} ({hrs:.1f} hrs)")
    print()
```
